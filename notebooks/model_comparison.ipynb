```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
results = {
    'Model': ['SVC', 'Decision Tree', 'Random Forest'],
    'Accuracy': [88.49646821392533, 87.00807265388497, 90.13622603430878],
    'Precision': [88.42828077314344, 92.57553956834532, 90.49541703793825],
    'Recall': [99.97124784358827, 92.60218767990789, 90.13622603430878],
    'F1 Score': [93.84615384615384, 92.5888617067204, 87.3066065774168]
}
results_df = pd.DataFrame(results)
results_df
```

```python
results_df.set_index('Model').plot(kind='bar', figsize=(12, 6))
plt.title('Model Performance Comparison')
plt.ylabel('Score (%)')
plt.xlabel('Model')
plt.ylim(0, 105)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
for metric in ['Accuracy', 'Precision', 'Recall', 'F1 Score']:
    row = results_df.loc[results_df[metric].idxmax()]
    print(f"Best {metric}: {row['Model']} ({row[metric]:.2f}%)")
```
