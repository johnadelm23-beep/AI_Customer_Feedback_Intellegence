```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Logistic Regression: Acc=93.11%, Prec=93.62%, Rec=98.88%, F1=96.18%
# Random Forest: Acc=91.52%, Prec=91.73%, Rec=99.28%, F1=95.36%
# Naive Bayes: Acc=90.77%, Prec=90.74%, Rec=99.63%, F1=94.98%
# SVC: Acc=88.75%, Prec=88.62%, Rec=100.00%, F1=93.97%
# Decision Tree: Acc=86.68%, Prec=92.77%, Rec=91.97%, F1=92.37%

results = {
    'Model': ['SVC', 'Decision Tree', 'Random Forest', 'logistic regression', 'naive bayes'],
    'Accuracy': [88.748739, 86.680121, 91.523713, 93.113017, 90.766902],
    'Precision': [88.622449, 92.770035, 91.728723, 93.622240, 90.744625],
    'Recall': [100.000000, 91.968912, 99.280368, 98.877375, 99.625792],
    'F1 Score': [93.968082, 92.367732, 95.355343, 96.178126, 94.978049]
}
results_df = pd.DataFrame(results)
results_df
```

```python
results_df.set_index('Model').plot(kind='bar', figsize=(12, 6))
plt.title('Model Performance Comparison')
plt.ylabel('Score (%)')
plt.xlabel('Model')
plt.ylim(0, 105)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
for metric in ['Accuracy', 'Precision', 'Recall', 'F1 Score']:
    row = results_df.loc[results_df[metric].idxmax()]
    print(f"Best {metric}: {row['Model']} ({row[metric]:.2f}%)")
```
