```python
import pandas as pd
from scipy.sparse import load_npz
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
```

```python
x_train = load_npz('X_train_tfidf.npz')
x_test = load_npz('X_test_tfidf.npz')

y_train = pd.read_csv('y_train.csv').values.ravel()
y_test = pd.read_csv('y_test.csv').values.ravel()

print('X_train:', x_train.shape)
print('X_test:', x_test.shape)
print('y_train:', y_train.shape)
print('y_test:', y_test.shape)
```

```python
decTree = DecisionTreeClassifier(random_state=42)

decTree.fit(x_train, y_train)
yPredict = decTree.predict(x_test)
```

```python
accuracy = accuracy_score(y_test, yPredict)
precision = precision_score(y_test, yPredict)
recall = recall_score(y_test, yPredict)
f1 = f1_score(y_test, yPredict)

print('Acc', accuracy * 100)
print('Precision', precision * 100)
print('Recall', recall * 100)
print('F1 score', f1 * 100)
```

```python
print(classification_report(y_test, yPredict))
```

```python
confusionMatrix = confusion_matrix(y_test, yPredict)
print(confusionMatrix)
```
