```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score    
from sklearn.metrics import  precision_score
from sklearn.metrics import   recall_score
from sklearn.metrics import   f1_score
from sklearn.metrics import    confusion_matrix
from sklearn.metrics import  classification_report
```

```python
dataSet = pd.read_csv('../data/raw/Womens Clothing E-Commerce Reviews.csv')

dataSet.head()
```

```python
copyDataSet=dataSet[['Review Text','Recommended IND']].copy()
copyDataSet.head()
```

```python
copyDataSet.isnull().sum()
```

```python
copyDataSet = copyDataSet.dropna(subset=['Review Text'])
copyDataSet.isnull().sum()
```

```python
copyDataSet.shape
```

```python
x=copyDataSet['Review Text']
y=copyDataSet['Recommended IND']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42, stratify=y)
print("Training:", x_train.shape)
print("Testing:", x_test.shape)
```

```python
TF_IDF=TfidfVectorizer(
    stop_words='english',
    max_features=5000,
)
xTrainTdIdf=TF_IDF.fit_transform(x_train)
xTestTdIdf=TF_IDF.transform(x_test)
print(xTrainTdIdf.shape)
print(xTestTdIdf)
```

```python
decTree=DecisionTreeClassifier(
    random_state=42
)
decTree.fit(xTrainTdIdf,y_train)
yPredict=decTree.predict(xTestTdIdf)
```

```python
accuracy = accuracy_score(y_test, yPredict)
precision = precision_score(y_test, yPredict)
recall = recall_score(y_test, yPredict)
f1 = f1_score(y_test, yPredict)

print("Accuracy:", accuracy*100)
print("Precision:", precision*100)
print("Recall:", recall*100)
print("F1 Score:", f1*100)
```

```python
print(classification_report(y_test, yPredict))
```

```python
confusionMatrix = confusion_matrix(y_test, yPredict)

print(confusionMatrix)
```
